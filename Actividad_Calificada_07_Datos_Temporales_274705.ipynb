## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# Escribe tu código aquí
# 1. Tipo de dato de pickup y dropoff
print("Tipos de dato originales:")
print(taxis[['pickup', 'dropoff']].dtypes)

# Valores faltantes antes de la conversion
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

# 2. Convertir a tipo datetime, formato (Ano-mes-dia Hora:minuto:segundo)
formato_fecha = '%Y-%m-%d %H:%M:%S'
taxis['pickup'] = pd.to_datetime(taxis['pickup'], format=formato_fecha)
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format=formato_fecha)

# 3. Verificar NaT generados
faltantes_despues = taxis[['pickup', 'dropoff']].isna().sum()

print("\nFaltantes antes de la conversión:\n", faltantes_antes)
print("\nFaltantes despues de la conversión:\n", faltantes_despues)
```

El 10 de marzo de 2019, en Nueva York, las horas entre 02:00 y 02:59 **no existieron**.

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# Escribe tu código aquí
# 1. Calcular la duracion
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# 2. Calcular la duracion en minutos
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

# 3. Resumen estadistico
print("Resumen estadistico de duracion_min:")
print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).
- Duración **mayor a 3 horas**.

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
# Escribe tu código aquí
# Contar anomalías usando máscaras booleanas
print("Duracion negativa:", (taxis['duracion_min'] < 0).sum())
print("Duracion cero:", (taxis['duracion_min'] == 0).sum())
print("Duracion menor a 1 min (incluyendo cero):", (taxis['duracion_min'] < 1).sum())
print("Duracion mayor a 3 horas (> 180 min):", (taxis['duracion_min'] > 180).sum())

# Mostrar viajes con duracion cero
cols = ['pickup', 'dropoff', 'distance', 'fare', 'total']
viajes_cero = taxis[taxis['duracion_min'] == 0][cols]
print("\nDetalle de viajes con duración cero:\n", viajes_cero)
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Escribe tu código aquí
# 1. Crear columna velocidad_mph (usando .where para dejar NaN si duracion <= 0)
velocidad_calculada = taxis['distance'] / (taxis['duracion_min'] / 60)
taxis['velocidad_mph'] = velocidad_calculada.where(taxis['duracion_min'] > 0)

# 2. Contar velocidades irreales
print("Viajes a más de 60 mph:", (taxis['velocidad_mph'] > 60).sum())

# 3. Top 5 viajes más rápidos
cols_vel = ['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']
top_5_rapidos = taxis.sort_values(by='velocidad_mph', ascending=False)[cols_vel].head(5)
print("\nTop 5 viajes más rápidos:\n", top_5_rapidos)
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí
# Creacion de columnas
taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6]) # 5 = Sábado, 6 = Domingo
taxis['fecha'] = taxis['pickup'].dt.date

# Viajes que cruzan la medianoche
taxis['cruza_medianoche'] = taxis['pickup'].dt.date != taxis['dropoff'].dt.date

# Suma
print("Total de viajes que cruzan la medianoche:", taxis['cruza_medianoche'].sum())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Escribe tu código aquí
# Agrupacion por hora para contar viajes y calcular mediana de duracion
patron_hora = taxis.groupby('hora').agg(
    num_viajes=('duracion_min', 'count'),
    mediana_duracion=('duracion_min', 'median')
)

print(patron_hora)
print("\nHora con mas viajes:", patron_hora['num_viajes'].idxmax())
print("Hora con menos viajes:", patron_hora['num_viajes'].idxmin())
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Escribe tu código aquí
# Agrupar por dia (numerico y nombre para mantener el orden de lunes a domingo)
patron_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    num_viajes=('duracion_min', 'count'),
    mediana_duracion=('duracion_min', 'median'),
    mediana_fare=('fare', 'median'),
    num_fechas_distintas=('fecha', 'nunique')
)

# Promedio de viajes dividiendo el total de viajes entre la cantidad de fechas del dia
patron_dia['promedio_viajes_por_fecha'] = patron_dia['num_viajes'] / patron_dia['num_fechas_distintas']

print(patron_dia[['num_viajes', 'mediana_duracion', 'mediana_fare', 'promedio_viajes_por_fecha']])
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Escribe tu código aquí
# Mascara de exclusion
m_marzo = (taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3)
m_duracion_valida = (taxis['duracion_min'] > 0)
m_velocidad_plausible = (taxis['velocidad_mph'] <= 60)

print("Excluidos por fecha (fuera de marzo):", (~m_marzo).sum())
print("Excluidos por duración (cero o negativa):", (~m_duracion_valida).sum())
print("Excluidos por velocidad (> 60 mph):", (~m_velocidad_plausible).sum())

# Mascara total
mascara_limpia = m_marzo & m_duracion_valida & m_velocidad_plausible

taxis_limpio = taxis[mascara_limpia].copy()
print("\nTotal de viajes excluidos en conjunto:", (~mascara_limpia).sum())
print("Dimensiones de taxis_limpio:", taxis_limpio.shape)

# Verificacion de que las anomalias se eliminaron
print("Velocidad maxima en limpio:", taxis_limpio['velocidad_mph'].max())
print("Duración minima en limpio:", taxis_limpio['duracion_min'].min())
```

### 4.2 Comparación entre periodos (6 pts)

Con `taxis_limpio`, compara los viajes que inician entre las **7:00 y las 9:59** en **días hábiles** contra los del **mismo horario en fin de semana**. Para cada grupo reporta: número de viajes, mediana de `duracion_min` y mediana de `velocidad_mph`.

```python
# Escribe tu código aquí
# Filtrar 7:00 a 9:59
m_horario = taxis_limpio['hora'].between(7, 9)

# Separar dias hábiles y fines de semana
habiles = taxis_limpio[m_horario & ~taxis_limpio['fin_semana']]
fines = taxis_limpio[m_horario & taxis_limpio['fin_semana']]

print("Horario 7:00 a 9:59 en dias habiles")
print("Viajes totales:", len(habiles))
print("Mediana de duración:", round(habiles['duracion_min'].median(), 2), "minutos")
print("Mediana de velocidad:", round(habiles['velocidad_mph'].median(), 2), "mph")

print("\nHorario 7:00 a 9:59 en Fin de Semana")
print("Viajes totales:", len(fines))
print("Mediana de duración:", round(fines['duracion_min'].median(), 2), "minutos")
print("Mediana de velocidad:", round(fines['velocidad_mph'].median(), 2), "mph")
```
