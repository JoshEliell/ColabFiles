**Práctica 2: Expresiones Regulares**

**Instrucciones:**

En esta practica, se te pide que diseñes un código en Google Colab para analizar el archivo informe_acciones.txt. El archivo contiene varios correos electrónicos en donde se informa el valor de las acciones de diferentes empresas. El código deberá extraer el valor actual de las acciones que aparecen en cada correo electrónico e imprimirlo a pantalla. En este ejercicio debeás utilizar Expresiones Regulares


**Ejemplo imagen de los valores que se busca extraer en los datos:**

image.png

```python
import re # Libreria expresiones regulares
from google.colab import drive
drive.mount('/content/drive')


ruta_datos = '/content/drive/MyDrive/ClassFiles/informe_acciones.txt'

valores_acciones = []

try:
  # encoding='latin-1' para evitar error al leer acentos
    with open(ruta_datos, 'r', encoding='latin-1') as archivo:
      # Iteramos linea por linea del documento para buscar nuestros datos
      print('Lista:')
      for linea in archivo:
        linea = linea.rstrip()
        # Se busca el patron utilizando re.search() Ejemplo (Valor actual: $960.36)
        coincidencia = re.search(r'Valor actual: \$([0-9.]+)', linea)
            # Buscamos el simbolo ($). Se usa antes (\) para interpratarlo literalmente ya que en Regex '$' significa el final de la línea
            # Los parentesis () crean un grupo de captura. Encuentra todo el patron, pero guarda solo lo que esta dentro de
            # Los corchetes [] se buscan cualquier numero del 0 al 9 y puntos decimales
            # El simbolo de + busca uno o más números o puntos juntos, hasta que no encuentre
        if coincidencia:
          valor_texto = coincidencia.group(1) # Se usa .group(1) para extraer solo la parte entre los parentesis (el numero) .group(0) = 'Valor actual: $960.36'
          valor_numerico = float(valor_texto)
          valores_acciones.append(valor_numerico)
          print(f"Valor extraido de la accion: ${valor_numerico}")
    print(f"\nNumero de valores en total: {len(valores_acciones)}")
except FileNotFoundError:
  print(f"\nError: No se encuentra el archivo en la ruta '{ruta_datos}'.")
```
