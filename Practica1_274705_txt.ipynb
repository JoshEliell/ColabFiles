**Práctica 1: Manejo de Archivos**

**Instrucciones:**

En esta practica, se te pide que diseñes un código en Google Colab para analizar el archivo mbox.txt (revisa el laboratorio 1). El código deberá extraer las mediciones contenidas en cada línea que contiene la cadena X-DSPAM-Confidence.


**Objetivos**


1. Tu código debe extraer únicamente los datos numéricos de las líneas que contienen la cadena X-DSPAM-Confidence en el archivo mbox.txt.
2. Para extraer los datos númericos usa las intrucciones vistas en el laboratorio 1 (startswith). Aquí no se usan expresiones regulares.
3. Los datos numéricos los deberás guardar en una lista y posteriormente mostrar los resultados de cada medición encontrada.
4. Explica brevemente en tu Google Colab qué realiza el código.
5. Calcular y mostrar en pantalla:
    *   La cantidad de datos extraídos.
    *   La sumatoria de los datos.
    *   El promedio de los datos.
    *   La varianza de los datos.

**Ejemplo imagen de los valores que se busca extraer en los datos:**

```python
#from google.colab import drive
#drive.mount('/content/drive')
lista = []

ruta_datos = '/content/drive/MyDrive/ClassFiles/mbox.txt'

try:
  with open(ruta_datos, 'r') as archivo:
    # Iteramos linea por linea del documento para buscar nuestros datos
    for linea in archivo:
      # Verificamos si la linea actual empieza con el patron que buscamos
      if linea.startswith('X-DSPAM-Confidence'):
        #Usamos split(':') para cortar el texto separando el texto del valor numerico
        partes = linea.split(':')
        #Partes[0] guarda "X-DSPAM-Confidence" y la partes[1] guardara" 0.8475"
        valor = float(partes[1].strip()) #Strip elimina cualquier espacio en blanco
        lista.append(valor)
        print(f"Valor de medicion encontrado: {valor}")
  cantidad = len(lista) #Cuenta la cantidad de valores en la lista
  if cantidad > 0:
    sumatoria = sum(lista) #El metodo sum suma los valores de la lista
    promedio = sumatoria / cantidad #El promedio es la suma total dividida entre la cantidad de datos
    #Formula de la varianza
    #Calculamos la diferencia entre cada numero y el promedio lo elevamos al cuadrado (** 2), y luego sumamos todos esos resultados
    suma_diferencias_cuadrado =sum((x - promedio) ** 2 for x in lista)
    varianza = suma_diferencias_cuadrado / cantidad
    print("\nRESULTADOS FINALES")
    print(f"Cantidad de datos extraidos : {cantidad}")
    print(f"Sumatoria de los datos      : {sumatoria}")
    print(f"Promedio de los datos       : {promedio}")
    print(f"Varianza de los datos       : {varianza}")
  else:
      print("\nNo se encontraron mediciones con la etiqueta 'X-DSPAM-Confidence:'.")
except FileNotFoundError:
  print(f"\nError: No se encuentra el archivo en la ruta '{ruta_datos}'.")
```
